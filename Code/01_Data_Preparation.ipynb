```python
import os
import random

import numpy as np
import pandas as pd

from pathlib import Path
```

```python
# Configuration

PROJECT_NAME = "LLM Hallucination Detection"

SEED = 42
NUM_QUESTIONS = 200

RAW_DATA_PATH = "/content/TruthfulQA.csv"
PROCESSED_DATA_PATH = "/content/truthfulqa_subset.csv"

MODEL_NAME = None        # Will be updated in Notebook 2

OUTPUT_FOLDER = "/content/results"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)
```

```python
random.seed(SEED)
np.random.seed(SEED)
```

```python
def load_truthfulqa(path):

    df = pd.read_csv(path)

    print(f"Rows: {len(df)}")
    print(f"Columns: {len(df.columns)}")

    return df
```

```python
truthfulqa_df = load_truthfulqa(RAW_DATA_PATH)
```

```python
truthfulqa_df.head()
```

```python
truthfulqa_df.columns
```

```python
def create_question_subset(df, sample_size=200, seed=42):
    """
    Create a reproducible subset of TruthfulQA questions.

    Parameters:
        df (DataFrame): Original TruthfulQA dataset.
        sample_size (int): Number of questions to sample.
        seed (int): Random seed for reproducibility.

    Returns:
        DataFrame: Sampled subset.
    """

    subset = (
        df.sample(n=sample_size, random_state=seed)
          .reset_index(drop=True)
    )

    return subset
```

```python
truthfulqa_subset = create_question_subset(
    truthfulqa_df,
    sample_size=NUM_QUESTIONS,
    seed=SEED
)

truthfulqa_subset.insert(0, "Question_ID", range(1, len(truthfulqa_subset) + 1))

print(f"Subset Size: {len(truthfulqa_subset)}")
truthfulqa_subset.head()
```

```python
truthfulqa_subset.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(f"Saved to: {PROCESSED_DATA_PATH}")
```
